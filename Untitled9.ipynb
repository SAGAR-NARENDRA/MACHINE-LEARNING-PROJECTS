# **1.Which states showed the highest case growth rates during peak months?**

```python
# Convert Date column to datetime
covid_data['Date'] = pd.to_datetime(covid_data['Date'], format='%Y-%m-%d')

# Extract month and year for peak period identification
covid_data['Month'] = covid_data['Date'].dt.month
covid_data['Year'] = covid_data['Date'].dt.year

# Aggregate confirmed cases by state and date
state_daily_cases = covid_data.groupby(['Date', 'State'])['Confirmed'].sum().reset_index()

# Calculate daily new cases
state_daily_cases['NewCases'] = state_daily_cases.groupby('State')['Confirmed'].diff().fillna(0)

# Identify peak months by summing new cases across India
monthly_cases = state_daily_cases.groupby([state_daily_cases['Date'].dt.to_period("M")])['NewCases'].sum()
peak_months = monthly_cases.sort_values(ascending=False).head(3).index

# Filter for peak months
state_daily_cases['MonthPeriod'] = state_daily_cases['Date'].dt.to_period("M")
peak_data = state_daily_cases[state_daily_cases['MonthPeriod'].isin(peak_months)]

# Calculate growth rates for each state during these months
peak_growth = peak_data.groupby(['State', 'MonthPeriod'])['NewCases'].sum().reset_index()

# Sort by highest growth
top_growth_states = peak_growth.sort_values(by='NewCases', ascending=False).groupby('MonthPeriod').head(5)

top_growth_states
```

# **2.Was there a significant difference in mortality rates between the first and second wave?**


```python
# Define wave periods
# First wave: March 2020 to November 2020
# Second wave: March 2021 to June 2021

first_wave = (covid_data['Date'] >= '2020-03-01') & (covid_data['Date'] <= '2020-11-30')
second_wave = (covid_data['Date'] >= '2021-03-01') & (covid_data['Date'] <= '2021-06-30')

# Aggregate data for each wave
wave_data = {
    'First Wave': covid_data[first_wave],
    'Second Wave': covid_data[second_wave]
}

mortality_summary = []

for wave_name, data in wave_data.items():
    grouped = data.groupby('State').agg({
        'Deaths': 'max',
        'Confirmed': 'max'
    }).reset_index()
    grouped['MortalityRate (%)'] = (grouped['Deaths'] / grouped['Confirmed']) * 100
    grouped['Wave'] = wave_name
    mortality_summary.append(grouped)

# Combine both waves into one DataFrame
mortality_comparison = pd.concat(mortality_summary)

# Pivot to compare waves side-by-side
mortality_pivot = mortality_comparison.pivot(index='State', columns='Wave', values='MortalityRate (%)')
mortality_pivot['Change (%)'] = mortality_pivot['Second Wave'] - mortality_pivot['First Wave']

# Sort by largest change
mortality_pivot_sorted = mortality_pivot.sort_values(by='Change (%)', ascending=False)

mortality_pivot_sorted.head(10)
```

# **4. Does the relationship between testing rates and detected cases change over time?**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load testing data
# The format string here needs to match the actual format in the file.
# Based on the original code, it seems the format is already '%Y-%m-%d' from a previous cell,
# but the error cell had '%d/%m/%Y'. Let's assume it's '%Y-%m-%d' from the preprocessing steps.
# If the original file format was '%d/%m/%Y', the first pd.to_datetime call should use that.
# We will use the format used in previous cells which was '%Y-%m-%d' after initial cleaning.
testing_data['Date'] = pd.to_datetime(testing_data['Date']) # Assuming it's already processed to YYYY-MM-DD format

# Remove missing or zero test records
testing_data = testing_data.dropna(subset=['TotalSamples', 'Positive'])
testing_data = testing_data[(testing_data['TotalSamples'] > 0) & (testing_data['Positive'] > 0)]

# Extract month for time-based grouping
testing_data['Month'] = testing_data['Date'].dt.to_period('M')

# Calculate correlation for each month
monthly_correlations = (
    testing_data.groupby('Month')
    .apply(lambda x: x['TotalSamples'].corr(x['Positive']))
    .reset_index(name='Correlation')
)

# Convert the 'Month' Period objects to strings for plotting
monthly_correlations['Month_str'] = monthly_correlations['Month'].astype(str)

# Plot correlation trend over time
plt.figure(figsize=(12, 6))
# Use the new string column for the x-axis
sns.lineplot(data=monthly_correlations, x='Month_str', y='Correlation', marker='o')
plt.title('Correlation Between Total Tests and Positive Cases Over Time')
plt.xlabel('Month')
plt.ylabel('Pearson Correlation')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# **5. Which 2 interventions most effectively reduced transmission?**

```python
# Assume interventions defined manually:
interventions = {
    'Lockdown 1.0': ('2020-03-25', '2020-05-31'),
    'Vaccination Rollout': ('2021-01-16', '2021-03-31')
}

# Load and prepare COVID case data
covid_data['Date'] = pd.to_datetime(covid_data['Date'])
covid_data['NewCases'] = covid_data.groupby('State')['Confirmed'].diff().fillna(0)
covid_data['NewCases'] = covid_data['NewCases'].clip(lower=0)  # Remove negatives

# Analyze average daily cases before and after each intervention
intervention_effects = []

for name, (start, end) in interventions.items():
    pre_period = (covid_data['Date'] < pd.to_datetime(start))
    post_period = (covid_data['Date'] >= pd.to_datetime(start)) & (covid_data['Date'] <= pd.to_datetime(end))

    pre_cases = covid_data[pre_period].groupby('Date')['NewCases'].sum().mean()
    post_cases = covid_data[post_period].groupby('Date')['NewCases'].sum().mean()

    reduction = (pre_cases - post_cases) / pre_cases * 100 if pre_cases > 0 else 0
    intervention_effects.append((name, pre_cases, post_cases, round(reduction, 2)))

# Display results
for name, pre, post, reduction in intervention_effects:
    print(f"{name}: Avg Cases Before = {pre:.0f}, After = {post:.0f}, Reduction = {reduction}%")
```
